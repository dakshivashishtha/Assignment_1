# Sales Prediction System Using Machine Learning

**Objective:** Predict `Sales_Amount` from historical sales transaction data.

This notebook follows the internship assignment workflow:

1. Import Libraries
2. Load Dataset
3. Dataset Understanding
4. Data Preprocessing
5. Date Feature Extraction
6. Exploratory Data Analysis
7. Feature Engineering
8. Feature Selection
9. Train-Test Split
10. Categorical Encoding
11. Model Training
12. Model Evaluation
13. Model Comparison
14. Final Sales Prediction
15. Save Model & Conclusion

> The dataset described for this project contains transaction-level sales information such as `Sale_Date`, `Sales_Rep`, `Region`, `Sales_Amount`, `Quantity_Sold`, `Product_Category`, `Unit_Cost`, `Unit_Price`, `Customer_Type`, `Discount`, `Payment_Method`, and `Sales_Channel`.


```python
# 1. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import joblib
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
```

## 2. Load Dataset

```python
# Change this path if your CSV has a different name/location

file_path = "sales_data.csv"

df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)
df.head()
```

## 3. Dataset Understanding

```python
# Basic information

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
# Statistical summary

df.describe(include='all').T
```

```python
# Missing values and duplicates

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Unique values in important categorical columns

for col in ['Sales_Rep', 'Region', 'Product_Category',
            'Customer_Type', 'Payment_Method', 'Sales_Channel']:
    if col in df.columns:
        print(f"{col}: {df[col].nunique()} unique values")
        print(df[col].unique())
        print()
```

## 4. Data Preprocessing

```python
# Convert date column

df['Sale_Date'] = pd.to_datetime(df['Sale_Date'], errors='coerce')

# Convert numerical columns safely
numeric_columns = [
    'Sales_Amount',
    'Quantity_Sold',
    'Unit_Cost',
    'Unit_Price',
    'Discount'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df.dtypes)
```

```python
# Remove exact duplicate transactions

before = len(df)
df = df.drop_duplicates().copy()
after = len(df)

print("Duplicates removed:", before - after)
```

```python
# Check invalid values

print("Missing dates:", df['Sale_Date'].isna().sum())
print("Missing target values:", df['Sales_Amount'].isna().sum())

print("Negative Sales Amount:", (df['Sales_Amount'] < 0).sum())
print("Negative Quantity Sold:", (df['Quantity_Sold'] < 0).sum())
print("Negative Unit Cost:", (df['Unit_Cost'] < 0).sum())
print("Negative Unit Price:", (df['Unit_Price'] < 0).sum())
print("Discount outside 0-100%:", ((df['Discount'] < 0) | (df['Discount'] > 100)).sum())
```

```python
# Handle missing values

# Numerical columns: median imputation
for col in numeric_columns:
    if df[col].isna().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

# Categorical columns: mode imputation
categorical_columns = [
    'Sales_Rep',
    'Region',
    'Product_Category',
    'Customer_Type',
    'Payment_Method',
    'Sales_Channel'
]

for col in categorical_columns:
    if df[col].isna().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# Rows with an invalid/missing date cannot be used for date-based features
df = df.dropna(subset=['Sale_Date']).copy()

print(df.isnull().sum())
```

```python
# Remove clearly invalid target/transaction values

df = df[df['Sales_Amount'] >= 0].copy()
df = df[df['Quantity_Sold'] > 0].copy()
df = df[df['Unit_Cost'] >= 0].copy()
df = df[df['Unit_Price'] >= 0].copy()

print("Shape after preprocessing:", df.shape)
```

## 5. Date Feature Extraction

```python
df['Year'] = df['Sale_Date'].dt.year
df['Month'] = df['Sale_Date'].dt.month
df['Quarter'] = df['Sale_Date'].dt.quarter
df['Day'] = df['Sale_Date'].dt.day
df['Day_of_Week'] = df['Sale_Date'].dt.dayofweek
df['Week_of_Year'] = df['Sale_Date'].dt.isocalendar().week.astype(int)

df[['Sale_Date', 'Year', 'Month', 'Quarter',
    'Day', 'Day_of_Week', 'Week_of_Year']].head()
```

```python
# Sales by product category

category_sales = (
    df.groupby('Product_Category')['Sales_Amount']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
category_sales.plot(kind='bar')
plt.title('Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales Amount')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Quantity sold by month

monthly_quantity = (
    df.set_index('Sale_Date')['Quantity_Sold']
      .resample('ME')
      .sum()
)

plt.figure(figsize=(12, 5))
plt.plot(monthly_quantity.index, monthly_quantity.values)
plt.title('Monthly Quantity Sold')
plt.xlabel('Month')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Price vs Sales Amount

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Unit_Price',
    y='Sales_Amount',
    alpha=0.5
)
plt.title('Unit Price vs Sales Amount')
plt.xlabel('Unit Price')
plt.ylabel('Sales Amount')
plt.tight_layout()
plt.show()
```

```python
# Discount vs Sales Amount

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Discount',
    y='Sales_Amount',
    alpha=0.5
)
plt.title('Discount vs Sales Amount')
plt.xlabel('Discount (%)')
plt.ylabel('Sales Amount')
plt.tight_layout()
plt.show()
```

```python
# Sales by customer type

customer_sales = df.groupby('Customer_Type')['Sales_Amount'].sum()

plt.figure(figsize=(7, 5))
customer_sales.plot(kind='bar')
plt.title('Sales by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Sales channel comparison

channel_sales = df.groupby('Sales_Channel')['Sales_Amount'].sum()

plt.figure(figsize=(7, 5))
channel_sales.plot(kind='bar')
plt.title('Sales by Sales Channel')
plt.xlabel('Sales Channel')
plt.ylabel('Sales Amount')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Correlation between numerical variables

numeric_for_corr = [
    'Sales_Amount',
    'Quantity_Sold',
    'Unit_Cost',
    'Unit_Price',
    'Discount'
]

plt.figure(figsize=(8, 6))
sns.heatmap(
    df[numeric_for_corr].corr(),
    annot=True,
    fmt='.2f'
)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## 7. Feature Engineering

```python
# Additional business-related features

df['Total_Cost'] = df['Quantity_Sold'] * df['Unit_Cost']
df['Gross_Profit'] = df['Sales_Amount'] - df['Total_Cost']

# Price difference between selling price and cost
df['Unit_Margin'] = df['Unit_Price'] - df['Unit_Cost']

# Discounted unit price
df['Discounted_Unit_Price'] = (
    df['Unit_Price'] * (1 - df['Discount'] / 100)
)

df[['Total_Cost', 'Gross_Profit', 'Unit_Margin',
    'Discounted_Unit_Price']].head()
```

### Feature leakage check

`Sales_Amount` is the target, so features that are calculated directly from the target must **not** be used for prediction.

Therefore, `Gross_Profit` is created for business analysis but is excluded from the ML features because it directly uses `Sales_Amount`.


## 8. Feature Selection

```python
# Features used to predict Sales_Amount

features = [
    'Quantity_Sold',
    'Unit_Cost',
    'Unit_Price',
    'Discount',
    'Year',
    'Month',
    'Quarter',
    'Day',
    'Day_of_Week',
    'Week_of_Year',
    'Sales_Rep',
    'Region',
    'Product_Category',
    'Customer_Type',
    'Payment_Method',
    'Sales_Channel'
]

target = 'Sales_Amount'

X = df[features].copy()
y = df[target].copy()

print("Features:", features)
print("Target:", target)
```

## 9. Train-Test Split

```python
# Sort chronologically before splitting

model_data = df.sort_values('Sale_Date').copy()

X = model_data[features]
y = model_data[target]

# Chronological split: first 80% for training, last 20% for testing
split_index = int(len(model_data) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining period:")
print(model_data['Sale_Date'].iloc[0], "to",
      model_data['Sale_Date'].iloc[split_index - 1])

print("\nTesting period:")
print(model_data['Sale_Date'].iloc[split_index], "to",
      model_data['Sale_Date'].iloc[-1])
```

### Why chronological splitting?

Because the dataset contains `Sale_Date`, the model should be evaluated on later transactions rather than randomly mixing past and future observations. This better represents using historical sales information to predict future sales.


## 10. Categorical Encoding

```python
categorical_features = [
    'Sales_Rep',
    'Region',
    'Product_Category',
    'Customer_Type',
    'Payment_Method',
    'Sales_Channel'
]

numerical_features = [
    'Quantity_Sold',
    'Unit_Cost',
    'Unit_Price',
    'Discount',
    'Year',
    'Month',
    'Quarter',
    'Day',
    'Day_of_Week',
    'Week_of_Year'
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        ),
        (
            'numerical',
            'passthrough',
            numerical_features
        )
    ]
)
```

## 11. Model Training

```python
# Linear Regression

linear_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

linear_model.fit(X_train, y_train)
linear_predictions = linear_model.predict(X_test)
```

```python
# Decision Tree Regressor

tree_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', DecisionTreeRegressor(
        max_depth=10,
        random_state=42
    ))
])

tree_model.fit(X_train, y_train)
tree_predictions = tree_model.predict(X_test)
```

```python
# Random Forest Regressor

random_forest_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=100,
        max_depth=12,
        random_state=42,
        n_jobs=-1
    ))
])

random_forest_model.fit(X_train, y_train)
rf_predictions = random_forest_model.predict(X_test)
```

## 12. Model Evaluation

```python
def evaluate_model(model_name, y_true, predictions):
    mae = mean_absolute_error(y_true, predictions)
    mse = mean_squared_error(y_true, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, predictions)

    return {
        'Model': model_name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    }

linear_result = evaluate_model(
    'Linear Regression',
    y_test,
    linear_predictions
)

tree_result = evaluate_model(
    'Decision Tree',
    y_test,
    tree_predictions
)

rf_result = evaluate_model(
    'Random Forest',
    y_test,
    rf_predictions
)

linear_result, tree_result, rf_result
```

## 13. Model Comparison

```python
results = pd.DataFrame([
    linear_result,
    tree_result,
    rf_result
])

results
```

```python
# Compare RMSE

plt.figure(figsize=(9, 5))
sns.barplot(data=results, x='Model', y='RMSE')
plt.title('Model Comparison - RMSE')
plt.ylabel('RMSE')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Compare R2 Score

plt.figure(figsize=(9, 5))
sns.barplot(data=results, x='Model', y='R2 Score')
plt.title('Model Comparison - R² Score')
plt.ylabel('R² Score')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### Model selection

For this regression problem:

- Lower **MAE, MSE and RMSE** indicate smaller prediction errors.
- Higher **R²** indicates that the model explains more variation in the target.

The final model will be selected from the actual evaluation results rather than assuming a model is best beforehand.


```python
# Select the model with the lowest RMSE

best_model_name = results.loc[results['RMSE'].idxmin(), 'Model']

model_dictionary = {
    'Linear Regression': linear_model,
    'Decision Tree': tree_model,
    'Random Forest': random_forest_model
}

final_model = model_dictionary[best_model_name]

print("Selected model:", best_model_name)
```

## 14. Final Sales Prediction

```python
# Generate predictions using the selected model

final_predictions = final_model.predict(X_test)

prediction_results = model_data.iloc[split_index:][
    ['Product_ID', 'Sale_Date', 'Region', 'Product_Category']
].copy()

prediction_results['Actual_Sales'] = y_test.values
prediction_results['Predicted_Sales'] = final_predictions

prediction_results.head(10)
```

```python
# Overall actual vs predicted sales by date

plot_data = prediction_results.groupby('Sale_Date')[
    ['Actual_Sales', 'Predicted_Sales']
].sum()

plt.figure(figsize=(12, 5))

plt.plot(
    plot_data.index,
    plot_data['Actual_Sales'],
    label='Actual Sales'
)

plt.plot(
    plot_data.index,
    plot_data['Predicted_Sales'],
    label='Predicted Sales'
)

plt.title('Actual vs Predicted Sales')
plt.xlabel('Date')
plt.ylabel('Sales Amount')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Save prediction results

import os

os.makedirs('outputs', exist_ok=True)

prediction_results.to_csv(
    'outputs/final_predictions.csv',
    index=False
)

results.to_csv(
    'outputs/model_comparison.csv',
    index=False
)

print("Prediction and model comparison files saved.")
```

## 15. Save Model & Conclusion

```python
# Save the final trained model

os.makedirs('models', exist_ok=True)

joblib.dump(
    final_model,
    'models/sales_prediction_model.pkl'
)

print("Model saved successfully.")
```
