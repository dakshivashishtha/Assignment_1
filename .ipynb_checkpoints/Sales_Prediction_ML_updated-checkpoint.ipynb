# Sales Prediction System Using Machine Learning — Updated Version

This is an updated version of your notebook. Changes from the original are called out in
**"Update note"** blocks throughout. The two biggest changes:

1. **Fixed two crashing bugs** (`tree_model` → `decision_tree_model`, `model_data` → `df`).
2. **Removed data leakage.** The original features included `Quantity_Sold`, `Unit_Price`,
   `Discount`, and `Discounted_Unit_Price` — but `Sales_Amount` is *calculated directly* from
   those same values (`Quantity_Sold × Unit_Price × (1 - Discount)` + 5% noise). Feeding them
   into the model let it "predict" the target by nearly re-deriving the formula it was built
   from, which is why R² was ~0.99 for every model. Real sales forecasting doesn't get to see
   the exact quantity/price/discount of a transaction that hasn't happened yet — that's what
   you're trying to estimate. The model below is retrained on only what would realistically be
   known **in advance** (who's selling, where, to whom, through what channel, when, and recent
   sales history) — plus new non-leaky lag/rolling features that replace the leaky ones.

## 5. Date Feature Extraction

```python
df['Year'] = df['Sale_Date'].dt.year
df['Month'] = df['Sale_Date'].dt.month
df['Quarter'] = df['Sale_Date'].dt.quarter
df['Day'] = df['Sale_Date'].dt.day
df['Day_of_Week'] = df['Sale_Date'].dt.dayofweek
df['Week_of_Year'] = df['Sale_Date'].dt.isocalendar().week.astype(int)

df[['Sale_Date', 'Year', 'Month', 'Quarter', 'Day', 'Day_of_Week', 'Week_of_Year']].head()
```

**Update note:** Added cyclical (sin/cos) encoding for `Month` and `Day_of_Week`.
A raw integer encoding tells the model December (12) and January (1) are far apart, when
they're actually adjacent in the calendar. Cyclical encoding fixes that.


```python
df['Month_sin'] = np.sin(2 * np.pi * df['Month'] / 12)
df['Month_cos'] = np.cos(2 * np.pi * df['Month'] / 12)
df['DOW_sin'] = np.sin(2 * np.pi * df['Day_of_Week'] / 7)
df['DOW_cos'] = np.cos(2 * np.pi * df['Day_of_Week'] / 7)

df[['Month', 'Month_sin', 'Month_cos', 'Day_of_Week', 'DOW_sin', 'DOW_cos']].head()
```

```python
numeric_cols = ['Sales_Amount', 'Quantity_Sold', 'Unit_Cost', 'Unit_Price', 'Discount']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
    print(col, ":", len(outliers), "outliers")
```

## 7. Feature Engineering

**Update note — this section is substantially revised.**

The original notebook engineered `Total_Cost`, `Unit_Margin`, and `Discounted_Unit_Price` —
all of which are direct algebraic functions of `Quantity_Sold`, `Unit_Cost`, `Unit_Price`, and
`Discount`, the very inputs used to construct `Sales_Amount`. Using them as model features
lets the model "cheat" by reconstructing the target formula instead of learning real sales
patterns.

Kept below (for business analysis only, **not** as model inputs):
- `Total_Cost`, `Gross_Profit`, `Unit_Margin` — useful for profitability reporting.

Replaced with genuinely predictive, non-leaky features:
- **`Prev_Sales_Amount`** — the sales amount of the *previous* transaction in the same
  Region + Product Category (a real "previous sales" signal, computed only from the past).
- **`Rolling_Avg_Sales_3`** / **`Rolling_Avg_Sales_7`** — rolling average of the last 3 / 7
  transactions in that Region + Product Category, shifted by one so the current row's own
  value is never included (this is what prevents leakage in a rolling feature).
- **`Month_sin`/`Month_cos`/`DOW_sin`/`DOW_cos`** — cyclical date encodings (added in Section 5).


```python
# Business-only features (NOT used as model inputs — informational / for reporting)
df['Total_Cost'] = df['Quantity_Sold'] * df['Unit_Cost']
df['Gross_Profit'] = df['Sales_Amount'] - df['Total_Cost']
df['Unit_Margin'] = df['Unit_Price'] - df['Unit_Cost']

df[['Total_Cost', 'Gross_Profit', 'Unit_Margin']].head()
```

```python
# Sort chronologically WITHIN each Region + Product_Category group so lag/rolling
# features only ever look backward in time for that specific segment.
df = df.sort_values(['Region', 'Product_Category', 'Sale_Date']).reset_index(drop=True)

group_cols = ['Region', 'Product_Category']

# Previous transaction's sales amount within the same group
df['Prev_Sales_Amount'] = df.groupby(group_cols)['Sales_Amount'].shift(1)

# Rolling averages, shifted by 1 so the current row's own Sales_Amount is excluded
df['Rolling_Avg_Sales_3'] = (
    df.groupby(group_cols)['Sales_Amount']
      .transform(lambda s: s.shift(1).rolling(window=3, min_periods=1).mean())
)
df['Rolling_Avg_Sales_7'] = (
    df.groupby(group_cols)['Sales_Amount']
      .transform(lambda s: s.shift(1).rolling(window=7, min_periods=1).mean())
)

# The very first transaction in each group has no history yet - fill with the
# overall median Sales_Amount as a reasonable default
for col in ['Prev_Sales_Amount', 'Rolling_Avg_Sales_3', 'Rolling_Avg_Sales_7']:
    df[col] = df[col].fillna(df['Sales_Amount'].median())

df[['Region', 'Product_Category', 'Sale_Date', 'Sales_Amount',
    'Prev_Sales_Amount', 'Rolling_Avg_Sales_3', 'Rolling_Avg_Sales_7']].head(10)
```

```python
# Re-sort back to chronological order overall (needed for the train/test split later)
df = df.sort_values('Sale_Date').reset_index(drop=True)
df.head()
```

## 8. Feature Selection

**Update note:** `Quantity_Sold`, `Unit_Price`, `Discount`, `Discounted_Unit_Price`, and
`Total_Cost` are **removed** from the model's feature list — these are exactly the values
`Sales_Amount` was calculated from, so including them was the source of the leakage / the
unrealistically high R² (~0.99) in the original notebook. Everything below is knowable
*before* the sale happens.


```python
features = [
    # Date-based (cyclical + non-cyclical that don't need re-encoding)
    'Year',
    'Quarter',
    'Day',
    'Week_of_Year',
    'Month_sin',
    'Month_cos',
    'DOW_sin',
    'DOW_cos',

    # Non-leaky history-based features
    'Prev_Sales_Amount',
    'Rolling_Avg_Sales_3',
    'Rolling_Avg_Sales_7',

    # Categorical business context
    'Sales_Rep',
    'Region',
    'Product_Category',
    'Customer_Type',
    'Payment_Method',
    'Sales_Channel'
]

target = 'Sales_Amount'

X = df[features].copy()
y = df[target].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 9. Train-Test Split

```python
# Sort data chronologically (already sorted, but re-confirm)
df = df.sort_values('Sale_Date').reset_index(drop=True)

X = df[features].copy()
y = df[target].copy()

# 80-20 chronological train-test split
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

### Why chronological splitting?

Because the dataset contains `Sale_Date`, the model should be evaluated on later
transactions rather than randomly mixing past and future observations. This better
represents using historical sales information to predict future sales.


## 10. Categorical Encoding

**Update note:** Added `drop='first'` to `OneHotEncoder`. Without it, one-hot encoded
columns are perfectly collinear (they always sum to 1), which inflates coefficient variance
in Linear Regression. Dropping the first category per column removes the redundancy without
losing any information — it doesn't affect tree-based models either way.


```python
categorical_features = [
    'Sales_Rep',
    'Region',
    'Product_Category',
    'Customer_Type',
    'Payment_Method',
    'Sales_Channel'
]

numerical_features = [
    'Year',
    'Quarter',
    'Day',
    'Week_of_Year',
    'Month_sin',
    'Month_cos',
    'DOW_sin',
    'DOW_cos',
    'Prev_Sales_Amount',
    'Rolling_Avg_Sales_3',
    'Rolling_Avg_Sales_7'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('categorical',
         OneHotEncoder(drop='first', handle_unknown='ignore'),
         categorical_features),

        ('numerical',
         'passthrough',
         numerical_features)
    ]
)
```

## 11. Model Training

```python
linear_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

linear_model.fit(X_train, y_train)
linear_predictions = linear_model.predict(X_test)

print("Linear Regression completed.")
```

```python
decision_tree_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', DecisionTreeRegressor(random_state=42))
])

decision_tree_model.fit(X_train, y_train)
decision_tree_predictions = decision_tree_model.predict(X_test)

print("Decision Tree completed.")
```

```python
random_forest_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=150,
        max_depth=12,
        random_state=42
    ))
])

random_forest_model.fit(X_train, y_train)
random_forest_predictions = random_forest_model.predict(X_test)

print("Random Forest completed.")
```

### Update: Gradient Boosting (added)

Gradient Boosting often outperforms a plain Random Forest on tabular regression tasks
because it builds trees sequentially, each one correcting the errors of the previous ones.


```python
gradient_boosting_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', GradientBoostingRegressor(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        random_state=42
    ))
])

gradient_boosting_model.fit(X_train, y_train)
gradient_boosting_predictions = gradient_boosting_model.predict(X_test)

print("Gradient Boosting completed.")
```

### Update: Hyperparameter Tuning (Random Forest)

Using `RandomizedSearchCV` with `TimeSeriesSplit` (instead of ordinary K-Fold) so that every
cross-validation fold still respects chronological order — never validating on data that's
older than what the fold trained on.


```python
param_distributions = {
    'model__n_estimators': [100, 150, 200, 300],
    'model__max_depth': [6, 8, 10, 12, None],
    'model__min_samples_leaf': [1, 2, 5, 10],
    'model__max_features': ['sqrt', 'log2', None]
}

tscv = TimeSeriesSplit(n_splits=5)

rf_search = RandomizedSearchCV(
    estimator=Pipeline([
        ('preprocessor', preprocessor),
        ('model', RandomForestRegressor(random_state=42))
    ]),
    param_distributions=param_distributions,
    n_iter=20,
    cv=tscv,
    scoring='neg_root_mean_squared_error',
    random_state=42,
    n_jobs=-1
)

rf_search.fit(X_train, y_train)

print("Best params:", rf_search.best_params_)
print("Best CV RMSE:", -rf_search.best_score_)

tuned_random_forest_model = rf_search.best_estimator_
tuned_rf_predictions = tuned_random_forest_model.predict(X_test)
```

## 12. Model Evaluation

```python
def evaluate_model(model_name, y_true, predictions):
    mae = mean_absolute_error(y_true, predictions)
    mse = mean_squared_error(y_true, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, predictions)

    return {
        'Model': model_name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    }

linear_result = evaluate_model('Linear Regression', y_test, linear_predictions)
tree_result = evaluate_model('Decision Tree', y_test, decision_tree_predictions)
rf_result = evaluate_model('Random Forest', y_test, random_forest_predictions)
gb_result = evaluate_model('Gradient Boosting', y_test, gradient_boosting_predictions)
tuned_rf_result = evaluate_model('Random Forest (Tuned)', y_test, tuned_rf_predictions)

linear_result, tree_result, rf_result, gb_result, tuned_rf_result
```

## 13. Model Comparison

```python
results = pd.DataFrame([
    linear_result,
    tree_result,
    rf_result,
    gb_result,
    tuned_rf_result
])

results
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=results, x='Model', y='RMSE')
plt.title('Model Comparison - RMSE (lower is better)')
plt.ylabel('RMSE')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=results, x='Model', y='R2 Score')
plt.title('Model Comparison - R² Score (higher is better)')
plt.ylabel('R2 Score')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### Model selection

For this regression problem:

- Lower **MAE, MSE and RMSE** indicate smaller prediction errors.
- Higher **R²** indicates that the model explains more variation in the target.

**Update note:** With the leaky features removed, expect R² to be noticeably lower than the
original notebook's ~0.99 — that's expected and correct. It now reflects genuine predictive
difficulty rather than the model reconstructing an arithmetic formula. The final model is
still selected from the actual evaluation results rather than assumed in advance.


```python
# Select the model with the lowest RMSE
best_model_name = results.loc[results['RMSE'].idxmin(), 'Model']

model_dictionary = {
    'Linear Regression': linear_model,
    'Decision Tree': decision_tree_model,
    'Random Forest': random_forest_model,
    'Gradient Boosting': gradient_boosting_model,
    'Random Forest (Tuned)': tuned_random_forest_model
}

final_model = model_dictionary[best_model_name]

print("Selected model:", best_model_name)
```

## 14. Final Sales Prediction

```python
# Generate predictions using the selected model
final_predictions = final_model.predict(X_test)

# Update: fixed bug - was 'model_data' (undefined), now correctly uses 'df'
prediction_results = df.iloc[split_index:][
    ['Product_ID', 'Sale_Date', 'Region', 'Product_Category']
].copy()

prediction_results['Actual_Sales'] = y_test.values
prediction_results['Predicted_Sales'] = final_predictions

prediction_results.head(10)
```

```python
# Overall actual vs predicted sales by date
plot_data = prediction_results.groupby('Sale_Date')[
    ['Actual_Sales', 'Predicted_Sales']
].sum()

plt.figure(figsize=(12, 5))
plt.plot(plot_data.index, plot_data['Actual_Sales'], label='Actual Sales')
plt.plot(plot_data.index, plot_data['Predicted_Sales'], label='Predicted Sales')
plt.title('Actual vs Predicted Sales')
plt.xlabel('Date')
plt.ylabel('Sales Amount')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
